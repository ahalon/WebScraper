# Packages import

```python
import requests
from bs4 import BeautifulSoup
```

# Ceneo Scraper

1. Provide url of the prodcut ' s opinions page.

```python
product_code = "84869312"
url = f'https://www.ceneo.pl/{product_code}#tab-reviews'
```

2. Send request to provided URL

```python
response = requests.get(url)
print(response.status_code)
```

3. Fetch product name

```python
page_dom = BeautifulSoup(response.text, "html.parser")
print(type(page_dom))
```

```python
product_name = page_dom.select_one('h1').get_text()
print(product_name)
print(type(product_name))
```

4. Fetch all opinions from the webpage

```python
opinions = page_dom.find_all("div", class_='js_product-review')
print(type(opinions))
print(len(opinions))
```

```python
opinions = page_dom.select("div.js_product-review")
print(type(opinions))
print(len(opinions))
```

5. Parse opinions to extract required data

```python
all_opinions = []

for opinion in opinions:
    single_opinion = {
        "opinion_id": opinion.get("data-entry-id"),
        "author": opinion.select("span.user-post__author-name")[0].get_text().strip() if opinion.select("span.user-post__author-name") else None,
        "recommendation": opinion.select("span.user-post__author-recomendation > em")[0].get_text().strip() if opinion.select("span.user-post__author-recomendation > em") else None,
        "score": opinion.select("span.user-post__score-count")[0].get_text().strip() if opinion.select("span.user-post__score-count") else None,
        "content": opinion.select("div.user-post__text")[0].get_text().strip() if opinion.select("div.user-post__text") else None,
        "pros": [tag.get_text().strip() for tag in opinion.select("div.review-feature__item--positive")],
        "cons": [tag.get_text().strip() for tag in opinion.select("div.review-feature__item--negative")],
        "helpful": opinion.select("button.vote-yes > span")[0].get_text().strip() if opinion.select("button.vote-yes > span") else "0",
        "unhelpful": opinion.select("button.vote-no > span")[0].get_text().strip() if opinion.select("button.vote-no > span") else "0",
        "publish_date": opinion.select("span.user-post__published > time:nth-child(1)")[0].get("datetime") if opinion.select("span.user-post__published > time:nth-child(1)") else None,
        "purchase_date": opinion.select("span.user-post__published > time:nth-child(2)")[0].get("datetime") if opinion.select("span.user-post__published > time:nth-child(2)") else None
    }
    all_opinions.append(single_opinion)

print(all_opinions)
```
