##  Packages import

```python
import os
import pandas as pd
from icalendar import Calendar, Event
```

## Calendar export

```python
group_name = input("Enter group name: ")
```

```python
classes = pd.read_csv(f"schedules/{group_name}.csv", encoding="UTF-8")
```

```python
if not os.path.exists("calendars"):
    os.mkdir("calendars")
```

```python
calendar = Calendar()
calendar["dtstart"] = "20260223T000000"
calendar["dtend"] = "20260930T235999"
calendar["summary"] = f"Schedule of the {group_name} group. "
```

```python
for index, row in classes.iterrows():
    event = Event()
    event["dtstart"] = "TZID=Europe/Warsaw:"+row["Termin"].replace("-", "")+row["Start time"].replace(":", "")+"00"
    event["dtend"] = "TZID=Europe/Warsaw:"+row["Termin"].replace("-", "")+row["End time"].replace(":", "")+"00"
    print(event.to_ical())
```
